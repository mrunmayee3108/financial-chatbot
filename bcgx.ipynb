```python
dff = pd.read_excel("Large_Financials_Dataset (1).xlsx")
```

Working on a copy of a dataset is always better, since it is possible that the original data may get altered.

```python
df = dff.copy()
```

Finding percentage change of revenue growth, net income growth and assets growth as compared to the previous year.

```python
df['Revenue Growth (%)'] = df.groupby(['Company'])['Total Revenue (USD millions)'].pct_change() * 100
```

```python
df['Net Income Growth (%)'] = df.groupby(['Company'])['Net Income (USD millions)'].pct_change() * 100
```

```python
df['Assets Growth (%)'] = df.groupby('Company')['Total Assets (USD millions)'].pct_change() * 100
```

Dropping the rows containing NaN

```python
df = df.dropna(subset=['Revenue Growth (%)', 'Net Income Growth (%)'])
```

Summary of revenue growth and net income growth of companies.

```python
company_summary = df.groupby('Company').agg({
    'Revenue Growth (%)': ['mean', 'min', 'max'],
    'Net Income Growth (%)': ['mean', 'min', 'max']
}).reset_index()
```

```python
company_summary
```

This shows that on average accross, Apple has the highest Revenue growth, and also the highest net income growth.

Summary of revenue growth and net income growth per year.

```python
year_summary = df.groupby('Fiscal Year').agg({
    'Revenue Growth (%)': 'mean',
    'Net Income Growth (%)': 'mean'
}).reset_index()
```

```python
year_summary
```

The revenue growth was highest in 2022, whereas the net income growth was highest in 2017.

Finding profit margin

```python
df['Profit Margin (%)'] = (df['Net Income (USD millions)'] / df['Total Revenue (USD millions)']) * 100
```

```python
df
```

```python
df.groupby('Fiscal Year')[['Fiscal Year', 'Profit Margin (%)']]
```

Ranking by total revenue and net income.

```python
df['Revenue rank'] = df.groupby('Fiscal Year')['Total Revenue (USD millions)'].rank(ascending=False)
```

```python
df['Net Income rank'] = df.groupby('Fiscal Year')['Net Income (USD millions)'].rank(ascending=False)
```

```python
df
```

```python
df[['Company', 'Fiscal Year', 'Revenue rank', 'Net Income rank']]
```

Finding volatility.

```python
volatility = df.groupby('Company')['Revenue Growth (%)'].std()
```

```python
volatility
```

Amazon is least volatile with a comparatively steady revenue growth. Tesla is the most volatile.

Finding the companies which had highest Revenue Growth in each year.

```python
df.loc[df.groupby('Fiscal Year')['Revenue Growth (%)'].idxmax(), ['Fiscal Year', 'Company', 'Revenue Growth (%)']].reset_index(drop = True)
```

Conclusion:

Apple has the highest Revenue growth, and also the highest net income growth.
Amazon has the most steadiest revenue growth, whereas Tesla has the least steady revenue growth.
The revenue growth was highest in 2022, whereas the net income growth was highest in 2017.
